Considerăm o urnă în care avem 3 bile roşii, 4 albastre şi 2 negre. Aruncăm un zar; dacă obţinem un număr prim,
adăugăm o bilă neagră în urnă, dacă obţinem 6, adăugăm o bilă roşie, iar în celelalte cazuri adăugăm o bilă albastră.
Apoi extragem o bilă din urnă.

a) Simulaţi în Python experimentul de mai sus.

```python
import random
def simuleaza_experiment():
    urna={"rosii":3, "albastre":4, "negre":2}
    zar=random.randint(1, 6)
    if zar in [2,3,5]:
        urna['negre'] += 1
    elif zar == 6:
        urna['rosii'] += 1
    else:
        urna['albastre'] += 1

    bile_in_urna = ( 
        ['rosie'] * urna['rosii'] +
        ['albastra'] * urna['albastre'] +
        ['neagra'] * urna['negre']
    )

    bila_extrasa = random.choice(bile_in_urna)
    return bila_extrasa

simuleaza_experiment()
```

b) Folosind modelarea făcută, estimaţi probabilitatea de a obţine o bilă roşie.

```python
nr_extrageri = 100000
extragere_rosie = 0

for i in range(nr_extrageri):
    bila = simuleaza_experiment()
    if bila == 'rosie':
        extragere_rosie += 1

print(extragere_rosie/nr_extrageri)
```
