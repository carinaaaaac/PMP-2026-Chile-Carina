# Exercițiul 1. De la independență condiționată la factorizare

Avem variabilele discrete $A, B, C$, graful neorientat $A - B - C$ și o distribuție comună strict pozitivă: $P(a,b,c) > 0$ pentru orice $(a,b,c)$.

## a) Ce înseamnă $A \perp C \mid B$?

Prin definiție, $A \perp C \mid B$ înseamnă că pentru orice valori $a, b, c$:
$$P(a, c \mid b) = P(a \mid b)\,P(c \mid b)\qquad\Longleftrightarrow\qquad P(a \mid b, c) = P(a \mid b).$$

Dacă știm deja valoarea lui $B$, atunci aflarea valorii lui $C$ nu mai aduce nicio informație suplimentară despre $A$ (și invers).
$A$ și $C$ pot fi dependente „în general” (influența trece prin $B$), dar această dependență dispare în momentul în care $B$ este fixat.
În graf, nodul $B$ separă $A$ de $C$: singurul drum de la $A$ la $C$ trece prin $B$.

## b) Pornind de la formula probabilităților condiționate, demonstrați că $A \perp C \mid B$ implică existența unor potențiale strict pozitive $\varphi_{AB}$ și $\varphi_{BC}$ și a unei constante $Z > 0$ pentru care
$$P(a,b,c)=\frac{1}{Z}\,\varphi_{AB}(a,b)\,\varphi_{BC}(b,c).$$

**Ipoteză:** $A \perp C \mid B$ și $P > 0$.

**Pasul 1 – cantitățile condiționate sunt bine definite și pozitive.**
$P(b)=\sum_{a,c}P(a,b,c)>0$ (sumă de termeni strict pozitivi), deci putem împărți la $P(b)$.
De asemenea $P(a,b)=\sum_c P(a,b,c)>0$, $P(b,c)=\sum_a P(a,b,c)>0$, deci $P(a\mid b)=\frac{P(a,b)}{P(b)}>0$ și $P(c\mid b)=\frac{P(b,c)}{P(b)}>0$.

**Pasul 2 – regula produsului și independența condiționată.**
$$P(a,b,c) \;=\; P(a,c\mid b)\,P(b) \;=\; P(a\mid b)\,P(c\mid b)\,P(b).$$
(prima egalitate este definiția probabilității condiționate, a doua folosește $A\perp C\mid B$.)

**Pasul 3 – alegerea potențialelor.** Grupăm factorii pe clicile grafului ($\{A,B\}$ și $\{B,C\}$):
$$\varphi_{AB}(a,b) := P(a\mid b) > 0,\qquad \varphi_{BC}(b,c) := P(c\mid b)\,P(b) = P(b,c) > 0,\qquad Z := 1 > 0 .$$
Atunci
$$\boxed{P(a,b,c)=\frac{1}{Z}\,\varphi_{AB}(a,b)\,\varphi_{BC}(b,c)}$$
cu potențiale strict pozitive și $Z>0$.

## c) Demonstrați și reciproca, pornind de la factorizarea de mai sus: dacă $$P(a,b,c)=\frac{1}{Z}\,\varphi_{AB}(a,b)\,\varphi_{BC}(b,c)$$ cu $\varphi_{AB},\varphi_{BC}>0$ și $Z>0$, atunci $A \perp C \mid B$.


**Ipoteză:** $P(a,b,c)=\frac1Z\varphi_{AB}(a,b)\varphi_{BC}(b,c)$ cu $\varphi_{AB},\varphi_{BC}>0$, $Z>0$.

Definim sumele parțiale (strict pozitive):
$$\alpha(b):=\sum_a \varphi_{AB}(a,b)>0,\qquad \beta(b):=\sum_c \varphi_{BC}(b,c)>0 .$$

**Pasul 1 – marginala lui $B$.** Suma pe $(a,c)$ se desparte în produs de două sume, deoarece $\varphi_{AB}$ depinde doar de $a$ (pentru $b$ fixat) iar $\varphi_{BC}$ doar de $c$:
$$P(b)=\sum_{a,c}\frac{1}{Z}\varphi_{AB}(a,b)\varphi_{BC}(b,c)=\frac1Z\Big(\sum_a\varphi_{AB}(a,b)\Big)\Big(\sum_c\varphi_{BC}(b,c)\Big)=\frac{\alpha(b)\,\beta(b)}{Z}>0.$$

**Pasul 2 – distribuția comună condiționată.**
$$\begin{aligned}
P(a,c\mid b)&=\frac{P(a,b,c)}{P(b)}=\frac{\dfrac{1}{Z}\,\varphi_{AB}(a,b)\,\varphi_{BC}(b,c)}{\dfrac{\alpha(b)\,\beta(b)}{Z}}\\
&=\frac{1}{Z}\,\varphi_{AB}(a,b)\,\varphi_{BC}(b,c)\cdot\frac{Z}{\alpha(b)\,\beta(b)}
=\frac{\varphi_{AB}(a,b)\,\varphi_{BC}(b,c)}{\alpha(b)\,\beta(b)}
=\frac{\varphi_{AB}(a,b)}{\alpha(b)}\cdot\frac{\varphi_{BC}(b,c)}{\beta(b)}.
\end{aligned}$$


**Pasul 3 – condiționatele individuale.** Însumând Pasul 2 după $c$, respectiv după $a$:
$$P(a\mid b)=\sum_c P(a,c\mid b)=\sum_c \frac{\varphi_{AB}(a,b)}{\alpha(b)}\cdot\frac{\varphi_{BC}(b,c)}{\beta(b)}=\frac{\varphi_{AB}(a,b)}{\alpha(b)}\underbrace{\sum_c\frac{\varphi_{BC}(b,c)}{\beta(b)}}_{=1}=\frac{\varphi_{AB}(a,b)}{\alpha(b)}$$

$$P(c\mid b)=\sum_a P(a,c\mid b)=\sum_a \frac{\varphi_{AB}(a,b)}{\alpha(b)}\cdot\frac{\varphi_{BC}(b,c)}{\beta(b)}=\frac{\varphi_{BC}(b,c)}{\beta(b)}\underbrace{\sum_a\frac{\varphi_{AB}(a,b)}{\alpha(b)}}_{=1}=\frac{\varphi_{BC}(b,c)}{\beta(b)}$$

Cele două sume sunt egale cu 1 deoarece $\sum_c \varphi_{BC}(b,c)=\beta(b)$ și $\sum_a \varphi_{AB}(a,b)=\alpha(b)$, conform definițiilor lui $\alpha$ și $\beta$.

**Concluzie.** Din Pasul 2 și Pasul 3:
$$P(a,c\mid b)=P(a\mid b)\,P(c\mid b)\quad\text{pentru orice }a,b,c,$$
adică $A\perp C\mid B$.




# Exercițiul 2. Putem prezice vremea la o stație fără observații?

**Datele alese**
- $n = 5$ stații: $S_1,\dots,S_5$, fiecare cu stările *Soare, Ploaie, Nori*.
- Muchiile (formează un arbore): $S_1-S_2,\; S_2-S_3,\; S_2-S_4,\; S_4-S_5$.
- Probabilități inițiale (aceleași pentru toate stațiile): $p_i=(0.5,\;0.3,\;0.2)$ pentru (Soare, Ploaie, Nori).
- Potențial pereche, același pentru toate muchiile (rânduri și coloane în ordinea Soare, Ploaie, Nori):
$$\Psi=\begin{pmatrix}0.7&0.2&0.1\\0.2&0.6&0.2\\0.1&0.3&0.6\end{pmatrix}.$$

> $\Psi$ este un **potențial** (o pondere), nu o matrice de probabilități condiționate: rândurile nu sunt normalizate separat; normalizarea se face o singură dată, global, prin constanta $Z$.
> Pentru muchia $(i,j)$ cu $i<j$ folosim mereu ordinea argumentelor $\Psi(s_i, s_j)$.

## a) Graful, vecinii și verificarea că nu are cicluri

```python
import numpy as np                       
import networkx as nx                    
import matplotlib.pyplot as plt          

STARI: list[str] = ["Soare", "Ploaie", "Nori"]          
N_STATII: int = 5
MUCHII: list[tuple[int, int]] = [(1, 2), (2, 3), (2, 4), (4, 5)]
P_INIT: list[float] = [0.5, 0.3, 0.2]                    
PSI: np.ndarray = np.array([[0.7, 0.2, 0.1],
                            [0.2, 0.6, 0.2],
                            [0.1, 0.3, 0.6]])

G = nx.Graph()
G.add_nodes_from(range(1, N_STATII + 1))
G.add_edges_from(MUCHII)

for i in G.nodes:
    print(f"Vecinii lui S{i}: {['S' + str(j) for j in sorted(G.neighbors(i))]}")

#Un graf conex cu n noduri si exact n-1 muchii este arbore (deci nu are cicluri)
print()
print("Număr noduri :", G.number_of_nodes(), "| număr muchii :", G.number_of_edges())
print("Graful este conex      :", nx.is_connected(G))
print("Graful nu are cicluri  :", nx.is_forest(G))
print("Graful este arbore     :", nx.is_tree(G))

poz = nx.spring_layout(G, seed=3)
plt.figure(figsize=(4.5, 3.2))
nx.draw(G, poz, with_labels=True, labels={i: f"S{i}" for i in G.nodes},
        node_color="lightskyblue", node_size=1100, font_size=12, edge_color="gray", width=2)
plt.title("Graful stațiilor meteo (arbore)")
plt.show()
```

## b) Modelul Markov discret în pyAgrum

Construim un `MarkovRandomField` cu 5 variabile discrete cu 3 stări. Adăugăm:
- câte un factor unar $\varphi_i(s)=p_i(s)$ pentru fiecare stație;
- câte un factor pereche $\varphi_{ij}(s_i,s_j)=\Psi_{s_i,s_j}$ pentru fiecare muchie $(i,j)$, mereu cu ordinea $(s_i, s_j)$, $i<j$.

Valorile se completează explicit, element cu element (prin dicționar `{variabilă: index}`), ca să nu depindem de ordinea internă a variabilelor din tensorul pyAgrum.

```python
import numpy as np                       
import pyagrum as gum                    

def construieste_mrf(n: int, muchii: list[tuple[int, int]], p: list[float], psi: np.ndarray
                     ) -> tuple[gum.MarkovRandomField, dict[int, gum.Tensor], dict[tuple[int, int], gum.Tensor]]:
    mrf = gum.MarkovRandomField()

    for i in range(1, n + 1):
        mrf.add(gum.LabelizedVariable(f"S{i}", f"Vremea la stația {i}", STARI))

    #factorii unari phi_i(s) = p_i(s)
    unari: dict[int, gum.Tensor] = {}
    for i in range(1, n + 1):
        f = mrf.addFactor({f"S{i}"})
        for s in range(3):
            f[{f"S{i}": s}] = p[s]
        unari[i] = f

    #factorii pereche phi_ij(si, sj) = Psi[si, sj]  (i < j)
    perechi: dict[tuple[int, int], gum.Tensor] = {}
    for (i, j) in muchii:
        f = mrf.addFactor({f"S{i}", f"S{j}"})
        for si in range(3):
            for sj in range(3):
                f[{f"S{i}": si, f"S{j}": sj}] = psi[si, sj]
        perechi[(i, j)] = f

    return mrf, unari, perechi


mrf, F_UNARI, F_PERECHE = construieste_mrf(N_STATII, MUCHII, P_INIT, PSI)


print("Factorul unar phi_1(s1):")
print(F_UNARI[1])
print("Factorul pereche phi_12(s1, s2) citit element cu element (rânduri = S1, coloane = S2) – coincide cu Psi:")
print(np.array([[F_PERECHE[(1, 2)][{"S1": a, "S2": b}] for b in range(3)] for a in range(3)]))
#afisarea interna a unui tensor pyAgrum poate arata tabelul transpus; valorile sunt aceleasi
```

## c) Graful modelului și distribuția comună

Graful modelului desenat de pyAgrum este un graf de factori: pătratele sunt variabilele $S_1,\dots,S_5$, iar cercurile mici sunt factorii. Cercurile legate de o singură variabilă sunt factorii unari $\varphi_i$, cele care leagă două variabile sunt factorii pereche $\varphi_{ij}$. Graful neorientat al dependențelor este arborele desenat la punctul a).

```python
import itertools                         
import pyagrum.lib.notebook as gnb       

gnb.showMRF(mrf)
```

**Distribuția comună** (Hammersley–Clifford, clicile unui arbore sunt nodurile și muchiile):
$$P(s_1,\dots,s_5)=\frac1Z\prod_{i=1}^{5}\varphi_i(s_i)\prod_{(i,j)\in E}\varphi_{ij}(s_i,s_j),$$
iar pentru graful ales:
$$P(s_1,\dots,s_5)=\frac1Z\;\underbrace{\varphi_1(s_1)\varphi_2(s_2)\varphi_3(s_3)\varphi_4(s_4)\varphi_5(s_5)}_{\text{factori unari}}\;\underbrace{\Psi(s_1,s_2)\,\Psi(s_2,s_3)\,\Psi(s_2,s_4)\,\Psi(s_4,s_5)}_{\text{factori pereche}},
\qquad Z=\sum_{s_1,\dots,s_5}(\ldots).$$

Cum avem doar $3^5=243$ configurații, putem calcula $Z$ direct și, mai departe, putem verifica orice rezultat al pyAgrum prin enumerare completă.

```python
def scor(config: dict[int, int]) -> float:
    #Produsul NENORMALIZAT al tuturor factorilor pentru o configuratie completa {i: stare_i}
    #P(config) = scor(config) / Z.
    val = 1.0
    for i, f in F_UNARI.items():
        val *= f[{f"S{i}": config[i]}]
    for (i, j), f in F_PERECHE.items():
        val *= f[{f"S{i}": config[i], f"S{j}": config[j]}]
    return val


#toate cele 3^5 configuratii ca dicționare {indice_stație: indice_stare}
TOATE_CONF: list[dict[int, int]] = [dict(zip(range(1, N_STATII + 1), t))
                               for t in itertools.product(range(3), repeat=N_STATII)]

Z = sum(scor(c) for c in TOATE_CONF)
print(f"Număr configurații: {len(TOATE_CONF)}")
print(f"Constanta de normalizare Z = {Z:.6f}")
print(f"Verificare: suma probabilităților = {sum(scor(c) / Z for c in TOATE_CONF):.6f}")
```

## d) Stații observate și marginala unei stații neobservate

Presupunem că vremea este cunoscută la două stații: $S_1=\text{Ploaie}$ și $S_5=\text{Nori}$.
Stațiile neobservate sunt $S_2, S_3, S_4$. Calculăm, cu inferență exactă (Shafer–Shenoy) din pyAgrum, distribuțiile marginale
$P(S_k\mid S_1=\text{Ploaie},S_5=\text{Nori})$ și alegem starea cea mai probabilă (**MAP marginal**).
Rezultatul este verificat independent prin enumerarea celor 243 de configurații.

```python
import numpy as np                       
import pyagrum as gum                    

OBSERVATII: dict[int, int] = {1: 1, 5: 2}           # S1 = Ploaie ,S5 = Nori
NEOBSERVATE: list[int] = [i for i in range(1, N_STATII + 1) if i not in OBSERVATII]
print("Observații:", {f"S{i}": STARI[s] for i, s in OBSERVATII.items()})
print("Neobservate:", [f"S{i}" for i in NEOBSERVATE])
print()

#inferenta exacta
ie = gum.ShaferShenoyMRFInference(mrf)
ie.setEvidence({f"S{i}": STARI[s] for i, s in OBSERVATII.items()})
ie.makeInference()

DIST_MARG: dict[int, list[float]] = {}
for i in NEOBSERVATE:
    post = ie.posterior(f"S{i}")
    DIST_MARG[i] = [post[{f"S{i}": s}] for s in range(3)]


# verificare prin enumerare
def configuratii_compatibile(obs: dict[int, int]) -> list[dict[int, int]]:
    #configuratiile care respectă observatiile
    return [c for c in TOATE_CONF if all(c[i] == s for i, s in obs.items())]


def marginale_enumerare(obs: dict[int, int]) -> dict[int, list[float]]:
    comp = configuratii_compatibile(obs)
    z_obs = sum(scor(c) for c in comp)                      
    rez = {i: [0.0] * 3 for i in range(1, N_STATII + 1) if i not in obs}
    for c in comp:
        for i in rez:
            rez[i][c[i]] += scor(c) / z_obs
    return rez


MARG_ENUM = marginale_enumerare(OBSERVATII)

for i in NEOBSERVATE:
    print(f"P(S{i} | observații):")
    for s, nume in enumerate(STARI):
        print(f"   {nume:<7} pyAgrum = {DIST_MARG[i][s]:.4f}   enumerare = {MARG_ENUM[i][s]:.4f}")
    mapm = int(np.argmax(DIST_MARG[i]))
    print(f"   -> starea cea mai probabilă la S{i} (MAP marginal): {STARI[mapm]}\n")

```

**Interpretare.** 
- **S2** este vecin direct cu $S_1$ (Ploaie): cea mai probabilă stare este **Ploaie** (0.53) – informația „trece” prin muchie.
- **S3** este legată de $S_1$ doar prin $S_2$: influența observației slăbește și, combinată cu preferința inițială $p_i$ (Soare = 0.5), starea cea mai probabilă devine **Soare** (0.47, față de Ploaie 0.39).
- **S4** este „trasă” în două direcții (de $S_2$ spre Ploaie, de $S_5$ spre Nori): distribuția este aproape plată (0.28 / 0.37 / 0.35), deci MAP marginal = **Ploaie**, dar cu incertitudine mare.
- Valorile din pyAgrum coincid cu cele din enumerare , deci modelul este construit corect.

## e) Configurația comună MAP și comparația cu maximele marginale

- **MAP marginal:** pentru fiecare stație separat, $\arg\max_{s_k}P(s_k\mid \text{obs})$.
- **MAP comun:** configurația *întreagă* $(s_2,s_3,s_4)^*=\arg\max_{s_2,s_3,s_4}P(s_2,s_3,s_4\mid\text{obs})$.

Deoarece $P(\cdot\mid \text{obs})\propto \text{scor}(\cdot)$ (constanta nu influențează maximul), căutăm maximul produsului de factori pe cele $3^3=27$ configurații compatibile cu observațiile.

```python
import numpy as np                      

comp = configuratii_compatibile(OBSERVATII)                
z_obs = sum(scor(c) for c in comp)

def prob_cond(c: dict[int, int]) -> float:
    # P(c | observatii) = scor(c)
    return scor(c) / z_obs

#MAP comun: configuratia cu scor maxim
map_comun = max(comp, key=scor)

#Configuratia formata din maximele marginale ( d) )
map_marginal = dict(OBSERVATII)
for i in NEOBSERVATE:
    map_marginal[i] = int(np.argmax(DIST_MARG[i]))

def text(c: dict[int, int]) -> str:
    return ", ".join(f"S{i}={STARI[c[i]]}" for i in range(1, N_STATII + 1))

print("MAP comun          :", text(map_comun), f"  -> P(config | obs) = {prob_cond(map_comun):.4f}")
print("Maxime marginale   :", text(map_marginal), f"  -> P(config | obs) = {prob_cond(map_marginal):.4f}")
print("Coincid?           :", map_comun == map_marginal)

print("\nPrimele 5 configurații după probabilitate:")
for c in sorted(comp, key=scor, reverse=True)[:5]:
    print("  ", text(c), f"  P = {prob_cond(c):.4f}")
```

**Interpretare.** 
**Nu coincid întotdeauna – și aici nu coincid.**
- MAP comun: $(S_2,S_3,S_4)=(\text{Soare},\text{Soare},\text{Soare})$, cu $P=0.162$.
- Maximele marginale: $(\text{Ploaie},\text{Soare},\text{Ploaie})$, cu $P=0.086$ (doar a 4-a configurație ca probabilitate, de aproape două ori mai puțin probabilă decât MAP-ul comun).

Motivul: maximul marginal al lui $S_2$ (Ploaie, 0.53) adună probabilitatea *mai multor* configurații în care $S_2$=Ploaie, pe când MAP-ul comun caută **o singură** configurație cu probabilitate maximă; aici configurația „toate Soare” este cea mai probabilă individual (potențialul $\Psi(\text{Soare},\text{Soare})=0.7$ este mare), iar probabilitatea Ploii este împărțită între mai multe configurații. Alegând separat maximele, putem obține o combinație care nu este nici ea cea mai probabilă globală (MAP este o problemă **globală** de optimizare).

# Exercițiul 3. Restaurarea unei imagini binare cu un MRF

## a) Imaginea originală $5\times5$

Pixel $0$ = negru, pixel $1$ = alb. Imaginea: un **pătrat alb $3\times3$** în centru, pe fond negru.

```python
import numpy as np                       
import matplotlib.pyplot as plt          

N: int = 5                                   

def arata_imagine(img: np.ndarray, ax, titlu: str, marcati: list[tuple[int, int]] | None = None) -> None:
    #marcati = pixeli încadrati cu rosu (modificati de zgomot)
    ax.imshow(img, cmap="gray", vmin=0, vmax=1)
    for r in range(img.shape[0]):
        for c in range(img.shape[1]):
            ax.text(c, r, str(img[r, c]), ha="center", va="center",
                    color="red" if img[r, c] == 0 else "blue", fontsize=11)
    for (r, c) in (marcati or []):
        ax.add_patch(plt.Rectangle((c - 0.5, r - 0.5), 1, 1, fill=False, edgecolor="orange", linewidth=3))
    ax.set_xticks(range(img.shape[1])); ax.set_yticks(range(img.shape[0]))
    ax.set_title(titlu)


x_orig = np.zeros((N, N), dtype=int)
x_orig[1:4, 1:4] = 1                         # patrat alb 3x3 în centru

print("Imaginea originală (matrice):")
print(x_orig)

fig, ax = plt.subplots(figsize=(3.2, 3.2))
arata_imagine(x_orig, ax, "Imagine originală")
plt.show()
```

## b) Zgomot: inversarea a ~10% dintre pixeli

Din 25 de pixeli, $10\%$ înseamnă $2{,}5$; alegem **3 pixeli** (12%) la întâmplare și îi inversăm ($0\leftrightarrow1$).
Folosim un generator cu **sămânță fixă** (`SEED = 9`) pentru reproductibilitate. Imaginea observată se notează $y$.

```python
import numpy as np                      
import matplotlib.pyplot as plt          

SEED: int = 9
K_ZGOMOT: int = 3                           

rng = np.random.default_rng(SEED)
idx = rng.choice(N * N, size=K_ZGOMOT, replace=False)        # indici (liniarizati) ai pixelilor inversati

y = x_orig.copy().ravel()
y[idx] = 1 - y[idx]                           # inversare 0 <-> 1
y = y.reshape(N, N)

pixeli_zgomot = [(int(i // N), int(i % N)) for i in idx]
print("Pixeli inversați (linie, coloană):", pixeli_zgomot)
print("Pixeli diferiți între y și original:", int((y != x_orig).sum()), "din", N * N)

fig, axs = plt.subplots(1, 2, figsize=(6.5, 3.4))
arata_imagine(x_orig, axs[0], "Originală")
arata_imagine(y, axs[1], "Observată y (cu zgomot)", marcati=pixeli_zgomot)
plt.show()
```

## c) Graful MRF și energia

**Variabilele.** Fiecare pixel necunoscut este o variabilă binară $x_i\in\{0,1\}$ ($i=0,\dots,24$, numerotare pe linii: $i=r\cdot5+c$).

**Graful.** Fiecare pixel este conectat doar cu vecinii direcți (sus, jos, stânga, dreapta): pixelii interiori au 4 vecini, cei de pe margine 3, colțurile 2.

**Energia.**
$$E(x;y)=\lambda\sum_i (x_i-y_i)^2+\sum_{(i,j)\in E}(x_i-x_j)^2,\qquad \lambda>0.$$

- **Termenul unar** $\lambda\sum_i(x_i-y_i)^2$ (*fidelitatea față de date*): costă $\lambda$ pentru fiecare pixel în care reconstrucția diferă de imaginea observată. Îl încurajează pe $x_i$ să rămână egal cu $y_i$.
- **Termenul pereche** $\sum_{(i,j)\in E}(x_i-x_j)^2$ (*regularitate / netezire*): costă $1$ pentru fiecare pereche de vecini cu valori diferite. Încurajează regiuni omogene.
- $\lambda$ controlează compromisul dintre cele două.

**Legătura cu probabilitățile (distribuția Gibbs):** $P(x\mid y)\propto e^{-E(x;y)}$, deci
$$P(x\mid y)\propto\prod_i\underbrace{e^{-\lambda(x_i-y_i)^2}}_{\varphi_i(x_i)}\;\prod_{(i,j)\in E}\underbrace{e^{-(x_i-x_j)^2}}_{\varphi_{ij}(x_i,x_j)}
\quad\Longrightarrow\quad \arg\max_x P(x\mid y)=\arg\min_x E(x;y).$$
(Energia **nu** este probabilitate: energie mică $\Rightarrow$ probabilitate mare, iar $Z$ nu influențează argmax-ul.)

```python
import numpy as np                       
import networkx as nx                    
import matplotlib.pyplot as plt         

def pixel(r: int, c: int) -> int:
    # Indexul liniarizat al pixelului (r, c).
    return r * N + c

MUCHII_GRID: list[tuple[int, int]] = []
for r in range(N):
    for c in range(N):
        if c + 1 < N:
            MUCHII_GRID.append((pixel(r, c), pixel(r, c + 1)))      # vecin la dreapta
        if r + 1 < N:
            MUCHII_GRID.append((pixel(r, c), pixel(r + 1, c)))      # vecin în jos

VECINI: dict[int, list[int]] = {p: [] for p in range(N * N)}
for (p, q) in MUCHII_GRID:
    VECINI[p].append(q)
    VECINI[q].append(p)

print("Număr muchii în grila 5x5:", len(MUCHII_GRID), "(= 2*5*4)")
print("Număr vecini: colț =", len(VECINI[pixel(0, 0)]),
      "| margine =", len(VECINI[pixel(0, 2)]),
      "| interior =", len(VECINI[pixel(2, 2)]))

Gg = nx.Graph(MUCHII_GRID)
poz_grid = {pixel(r, c): (c, -r) for r in range(N) for c in range(N)}
plt.figure(figsize=(4, 4))
nx.draw(Gg, poz_grid, with_labels=True, node_color="lightskyblue", node_size=450, font_size=9, edge_color="gray")
plt.title("Graful MRF: grilă 5x5, vecinătate de ordinul întâi")
plt.show()


def energie(x: np.ndarray, y: np.ndarray, lam: float) -> float:
    # E(x; y) = lam * sum_i (x_i - y_i)^2 + sum_{(i,j) in E} (x_i - x_j)^2
    xf, yf = x.ravel(), y.ravel()
    unar = lam * np.sum((xf - yf) ** 2)
    pereche = sum((xf[i] - xf[j]) ** 2 for (i, j) in MUCHII_GRID)
    return float(unar + pereche)

print("\nExemplu: E(y; y) cu lambda=1 =", energie(y, y, 1.0),
      "(doar termenul pereche, deoarece x = y)")
```

## d) Factorii în pyAgrum și configurația MAP (algoritmul ICM)

**Factorii.** Pentru fiecare pixel $i$: $\varphi_i(x_i)=e^{-\lambda(x_i-y_i)^2}$; pentru fiecare muchie: $\varphi_{ij}(x_i,x_j)=e^{-(x_i-x_j)^2}$.
Verificăm numeric că $-\log\prod\varphi=E(x;y)$ (adică factorii reprezintă exact energia).

**Algoritm: ICM (Iterated Conditional Modes).** Căutarea exactă a MAP pe 25 de variabile binare ar însemna $2^{25}\approx3{,}3\cdot10^7$ configurații; folosim ICM:

1. Pornim de la configurația inițială $x := y$ (imaginea observată).
2. Parcurgem pixelii pe rând (linie cu linie).
3. Pentru pixelul $i$, ținând fixe toate celelalte valori, alegem $x_i\in\{0,1\}$ care **maximizează** $\log\varphi_i(x_i)+\sum_{j\in N(i)}\log\varphi_{ij}(x_i,x_j)$, adică **minimizează energia locală**. (Doar factorii care conțin $x_i$ contează; restul sunt constanți.) La egalitate, păstrăm valoarea curentă.
4. Repetăm parcurgerea până când într-o trecere completă nu se mai schimbă niciun pixel.

Fiecare schimbare nu crește energia, deci algoritmul se oprește; **poate însă să se oprească într-un minim local**, nu neapărat în MAP-ul global.

```python
import itertools
import numpy as np
import pyagrum as gum


def construieste_mrf_imagine(y: np.ndarray, lam: float
                             ) -> tuple[gum.MarkovRandomField, dict[int, gum.Tensor], dict[tuple[int, int], gum.Tensor]]:
    # MRF pentru restaurare: 25 variabile binare, factori unari (cu y si lambda) si factori pereche
    mrf_img = gum.MarkovRandomField()
    for p in range(N * N):
        mrf_img.add(gum.LabelizedVariable(f"X{p}", f"pixelul {p}", ["0", "1"]))

    yf = y.ravel()
    unari: dict[int, gum.Tensor] = {}
    for p in range(N * N):
        f = mrf_img.addFactor({f"X{p}"})
        for v in (0, 1):
            f[{f"X{p}": v}] = np.exp(-lam * (v - yf[p]) ** 2)       # phi_i(x_i) = exp(-lambda (x_i - y_i)^2)
        unari[p] = f

    perechi: dict[tuple[int, int], gum.Tensor] = {}
    for (p, q) in MUCHII_GRID:
        f = mrf_img.addFactor({f"X{p}", f"X{q}"})
        for vp in (0, 1):
            for vq in (0, 1):
                f[{f"X{p}": vp, f"X{q}": vq}] = np.exp(-((vp - vq) ** 2))   # phi_ij = exp(-(x_i - x_j)^2)
        perechi[(p, q)] = f
    return mrf_img, unari, perechi



def icm(unari: dict[int, gum.Tensor], perechi: dict[tuple[int, int], gum.Tensor], y: np.ndarray
        ) -> tuple[np.ndarray, int]:
    # ICM: minimizare coordonata cu coordonata a energiei. Intoarce (imaginea reconstruita, nr. de treceri).
    x = y.copy().ravel()                                  # initializare: x = y

    def log_scor_local(p: int, v: int) -> float:
        # log phi_p(v) + suma log phi_pq(v, x_q) pe vecinii q (restul factorilor nu depind de x_p)
        s = np.log(unari[p][{f"X{p}": v}])
        for q in VECINI[p]:
            cheie = (p, q) if (p, q) in perechi else (q, p)                # muchia e stocata o singura data
            s += np.log(perechi[cheie][{f"X{p}": v, f"X{q}": int(x[q])}])
        return float(s)

    treceri = 0
    modificat = True
    while modificat:
        modificat = False
        treceri += 1
        for p in range(N * N):
            curent, alt = int(x[p]), 1 - int(x[p])
            if log_scor_local(p, alt) > log_scor_local(p, curent) + 1e-12:
                x[p] = alt
                modificat = True
    return x.reshape(N, N), treceri


def map_exact_pe_linii(y: np.ndarray, lam: float) -> tuple[np.ndarray, float]:
    # MAP EXACT prin programare dinamica pe linii (doar pentru verificare).
    # O linie are 5 pixeli => 2^5 = 32 stari posibile. Energia = costuri pe linie (unar + muchii orizontale)
    # + costuri intre linii consecutive (muchii verticale). Parcurgem liniile de sus in jos, pastrand
    # pentru fiecare stare a liniei curente cel mai mic cost posibil.
    linii = [np.array(t) for t in itertools.product((0, 1), repeat=N)]
    cost_linie = [[lam * np.sum((a - y[r]) ** 2) + np.sum((a[:-1] - a[1:]) ** 2) for a in linii] for r in range(N)]
    cost_trecere = np.array([[np.sum((a - b) ** 2) for b in linii] for a in linii])   # muchii verticale intre doua linii

    best = np.array(cost_linie[0])
    parinte: list[np.ndarray] = []
    for r in range(1, N):
        total = best[:, None] + cost_trecere                  # [linia de sus, linia curenta]
        parinte.append(total.argmin(axis=0))                  # cea mai buna linie de deasupra, pentru fiecare linie curenta
        best = total.min(axis=0) + np.array(cost_linie[r])

    k = int(best.argmin()) #indicele variantei cu cel mai mic cost
    e_min = float(best[k])
    idx_linii = [k] #de la ultima la prima
    for par in reversed(parinte):                             # reconstruim solutia de jos in sus
        k = int(par[k])
        idx_linii.append(k)
    x = np.array([linii[k] for k in reversed(idx_linii)])
    return x, e_min

```

```python
import numpy as np                       
import matplotlib.pyplot as plt          

LAMBDAS: list[float] = [0.3, 1.0, 2.0, 3.0]

rezultate: dict[float, np.ndarray] = {}
print(f"{'lambda':>6} | {'greșiți ICM':>11} | {'treceri':>7} | {'E(ICM)':>7} | {'E(MAP exact)':>12} | {'greșiți MAP exact':>17} | {'E(original)':>11}")
print("-" * 92)
print(f"{'y':>6} | {int((y != x_orig).sum()):>11} |   (imaginea zgomotoasă, fără restaurare)")

for lam in LAMBDAS:
    mrf_l, un_l, pr_l = construieste_mrf_imagine(y, lam)        # modelul pentru acest lambda
    x_rec, treceri = icm(un_l, pr_l, y)                         # MAP aproximativ prin ICM
    x_ex, e_ex = map_exact_pe_linii(y, lam)                     # MAP exact (verificare)
    rezultate[lam] = x_rec
    print(f"{lam:>6} | {int((x_rec != x_orig).sum()):>11} | {treceri:>7} | {energie(x_rec, y, lam):>7.2f} | "
          f"{e_ex:>12.2f} | {int((x_ex != x_orig).sum()):>17} | {energie(x_orig, y, lam):>11.2f}")

fig, axs = plt.subplots(1, 2 + len(LAMBDAS), figsize=(3.4 * (2 + len(LAMBDAS)), 3.8))
arata_imagine(x_orig, axs[0], "Originală")
arata_imagine(y, axs[1], f"Observată y\n({int((y != x_orig).sum())} greșiți)", marcati=pixeli_zgomot)
for ax, lam in zip(axs[2:], LAMBDAS):
    gres = [(r, c) for r in range(N) for c in range(N) if rezultate[lam][r, c] != x_orig[r, c]]
    arata_imagine(rezultate[lam], ax, f"ICM, λ = {lam}\n({len(gres)} greșiți)", marcati=gres)
plt.tight_layout()
plt.show()
```


**Rolul lui $\lambda$.** Termenul unar costă $\lambda$ pentru fiecare pixel schimbat față de $y$, cel pereche costă 1 pentru fiecare muchie între vecini diferiți.
- **$\lambda$ mic ($0{,}3$ și $1$) → netezire.** Fidelitatea față de date „costă puțin”, deci modelul acceptă să modifice mulți pixeli ca să elimine diferențele dintre vecini. Aici netezirea este atât de puternică încât **MAP-ul exact este imaginea complet neagră** (pătratul alb dispare: ștergerea lui costă $9\lambda$, iar păstrarea lui costă cele 12 muchii de pe contur, deci – aproximativ, ignorând pixelii de zgomot – pătratul este eliminat când $9\lambda<12$, adică $\lambda<4/3$).
- **$\lambda$ mare ($3$) → fidelitate.** Orice modificare a lui $y$ costă mult, iar zgomotul rămâne: un pixel izolat greșit cu $k$ vecini diferiți este corectat doar dacă $\lambda<k$ ($k=2$ în colț, $3$ pe margine, $4$ în interior). Pentru $\lambda=3$ rămân cei 3 pixeli de zgomot ($y$ nu este modificat).
- **Valoare intermediară ($\lambda=2$).** Pătratul este păstrat ($9\lambda>12$), iar zgomotul din colțuri începe să fie corectat: MAP-ul exact are un singur pixel greșit.

**ICM vs. MAP exact.** ICM coboară energia pas cu pas, dar se oprește în **minime locale**: pentru $\lambda=0{,}3$ și $\lambda=1$ energia găsită de ICM (7.40 și 15.00) este mai mare decât minimul global (3.60 și 12.00), deci ICM nu găsește MAP-ul. Rezultatul ICM cu un singur pixel greșit la $\lambda=1$ „arată bine”, dar nu minimizează energia. Pentru $\lambda=2$ și $3$ energiile coincid (17.00): minimul global nu este unic, iar ICM a găsit unul dintre minimizatori (la egalitate păstrează valoarea curentă).
