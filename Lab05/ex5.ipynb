# Exercițiu laborator 5: Modele Markov ascunse

Acest notebook conține rezolvarea problemei HMM utilizând librăria `hmmlearn`.

```python
import numpy as np
from hmmlearn import hmm

# 1. Definirea componentelor modelului
states = ["Dificil", "Mediu", "Ușor"]
n_states = len(states)

observations = ["FB", "B", "S", "NS"]
n_observations = len(observations)

# Probabilitățile inițiale
start_probability = np.array([1/3, 1/3, 1/3])

# Matricea de tranziție a stărilor (Dificil, Mediu, Ușor)
transition_probability = np.array([
    [0.0, 0.5, 0.5],   # Din Dificil
    [0.5, 0.25, 0.25], # Din Mediu
    [0.5, 0.25, 0.25]  # Din Ușor
])

# Matricea de emisie (FB, B, S, NS)
emission_probability = np.array([
    [0.10, 0.20, 0.40, 0.30], # Test Dificil
    [0.15, 0.25, 0.50, 0.10], # Test Mediu
    [0.20, 0.30, 0.40, 0.10]  # Test Ușor
])

# Inițializarea modelului
model = hmm.CategoricalHMM(n_components=n_states)
model.startprob_ = start_probability
model.transmat_ = transition_probability
model.emissionprob_ = emission_probability

# 2. Secvența observată de note
obs_map = {"FB": 0, "B": 1, "S": 2, "NS": 3}
obs_sequence = ["FB", "FB", "S", "B", "B", "S", "B", "B", "NS", "B", "B", "S"]
obs_indices = np.array([[obs_map[nota]] for nota in obs_sequence])

# 3. Determinarea celei mai probabile secvențe de dificultăți (Viterbi)
logprob, state_sequence = model.decode(obs_indices, algorithm="viterbi")

print("Secvența observată:", ", ".join(obs_sequence))
print("Secvența de stări (dificultăți):", ", ".join([states[state] for state in state_sequence]))
print(f"Log-Probabilitatea secvenței: {logprob}")
print(f"Probabilitatea secvenței: {np.exp(logprob)}")
```
